```python
from utils import iou_and_counts_detectron2, iou_and_counts_yolo
ious = []
detection_numbers = []
gt_sum = 0
TP = 0
FP = 0
FN = 0
with torch.inference_mode():
    for val_inputs in val_loader:
       if yolo_based:
            batch_images = torch.stack([val_input['image'] for val_input in val_inputs], dim=0)
            batch_outputs = predictor.predict_batch_tensor(batch_images)

            for val_input, outputs in zip(val_inputs, batch_outputs):
                pred_instances = outputs['instances'].to('cuda')
                gt_instances = val_input['instances'].to('cuda')
                metrics_dict = iou_and_counts_yolo(
                    pred_instances,
                    gt_instances,
                    iou_thresh=0.5,
                )

                ious.append(metrics_dict['mean_iou'])
                detection_numbers.append(metrics_dict['detected'])
                TP += metrics_dict['tp']
                FP += metrics_dict['fp']
                FN += metrics_dict['fn']
                gt_sum += len(val_input['instances'].gt_boxes)
       else:
            for val_input in val_inputs:
                outputs = predictor(val_input['image'].cpu().numpy().transpose((1, 2, 0)))
                metrics_dict = iou_and_counts_detectron2(
                    outputs['instances'].pred_boxes.to('cuda'),
                    outputs['instances'].scores,
                    val_input['instances'].gt_boxes.to('cuda'),
                    iou_thresh=0.5,
                )

                ious.append(metrics_dict['mean_iou'])
                detection_numbers.append(metrics_dict['detected'])
                TP += metrics_dict['tp']
                FP += metrics_dict['fp']
                FN += metrics_dict['fn']
                gt_sum += len(val_input['instances'].gt_boxes)

avg_iou = np.mean(ious)
asr = 1 - sum(detection_numbers) / gt_sum
Precision = TP / (TP + FP)
Recall = TP / (TP + FN)
F1_score = 2*(Precision * Recall)/(Precision + Recall)

print(f"Before Attack:")
print(f'F1 score: {F1_score}')
print(f'mIoU: {avg_iou}')
print(f'ASR: {asr}')
print(f'Precision: {Precision}')
print(f'Recall: {Recall}')
```
